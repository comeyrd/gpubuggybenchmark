```python
import dataLoader as dl
import basePlots as bp
import combinedPlots as cbp
import matplotlib.pyplot as plt
```

```python
EXPERIMENT_14_11_file = "../src/build/csv/experiment-14-11-2025"
exp_14_11 =  dl.Experiment(EXPERIMENT_14_11_file)
filter_14_11 = dl.Filter(exp_14_11.inner_df)
```

```python
filter_14_11.set_x("repetitions")
filter_14_11.set_y("repetitions_duration",explode=True)
filter_14_11.set_hue("repetitions",True)
filter_14_11.set_default("version","Reference")
filter_14_11.set_subplot("blocking",False)
filter_14_11.set_default("flush_l2",1)
filter_14_11.set_default("kernel","Bilateral")
filter_14_11.set_default("repetitions",500)
filter_14_11.set_default("blocking",1)
#filter_14_11.set_subsets("version",["RCAccuracy","MOAccuracy","UOAccuracy","ReferenceAccuracy"])
#filter_14_11.set_subsets("repetitions",[10,50,750])
```

```python
bp.compare_cdf(exp_14_11,filter_14_11)
sub_height = bp.subtitle(filter_14_11)
bp.do_title("Empirical Cumulative Distribution Function of execution times, comparing numbers of repetitions",sub_height,60)
plt.ylabel("Probability (D)")
plt.xlabel("Execution time (ms)")
plt.savefig("ecdf-progress-report.svg", bbox_inches='tight')
```

```python
in_filter = filter_14_11.copy()
in_filter.set_y("repetitions_duration",explode=True)
bp.compare_densities(exp_14_11,in_filter) 
sub_height = bp.subtitle(in_filter)
bp.do_title("Density of execution times, comparing numbers of repetitions",sub_height,55)
plt.xlabel("Execution time (ms)")
plt.ylabel("Density")
plt.savefig("density-progress-report.svg", bbox_inches='tight')
```
